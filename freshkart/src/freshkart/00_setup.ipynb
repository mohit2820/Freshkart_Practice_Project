## Widgets

```python
dbutils.widgets.text("catalog" , "fk_1393")
catalog = dbutils.widgets.get("catalog")
```

## Catalog

```python
%sql
CREATE CATALOG IF NOT EXISTS IDENTIFIER(:catalog);
```

## Schema

```python
%sql
CREATE SCHEMA IF NOT EXISTS IDENTIFIER(:catalog).landing;
CREATE SCHEMA IF NOT EXISTS IDENTIFIER(:catalog).data_quality;
CREATE SCHEMA IF NOT EXISTS IDENTIFIER(:catalog).ops;
CREATE SCHEMA IF NOT EXISTS IDENTIFIER(:catalog).bronze;
CREATE SCHEMA IF NOT EXISTS IDENTIFIER(:catalog).silver;
CREATE SCHEMA IF NOT EXISTS IDENTIFIER(:catalog).gold;
```

## Volume

```python
%sql
CREATE VOLUME IF NOT EXISTS IDENTIFIER(:catalog).landing.raw;
CREATE VOLUME IF NOT EXISTS IDENTIFIER(:catalog).ops.pipeline_state;
```

## Directories

```python
catalog = catalog.strip()
dbutils.fs.mkdirs(f"/Volumes/{catalog}/landing/raw/customers")
dbutils.fs.mkdirs(f"/Volumes/{catalog}/landing/raw/orders")
dbutils.fs.mkdirs(f"/Volumes/{catalog}/landing/raw/products")
dbutils.fs.mkdirs(f"/Volumes/{catalog}/landing/raw/stores")
```
